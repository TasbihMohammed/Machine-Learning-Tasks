## **Scaling**

```python
# Upload dataset file:
from google.colab import files

uploaded = files.upload()
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler

# Load the dataset
df = pd.read_csv("Scaling_data.csv")

# Select the feature that needs scaling
feature = df[["LotArea"]]
```

```python
# Apply Standard Scaling
standard_scaler = StandardScaler()
df_standard = df.copy()
df_standard["LotArea"] = standard_scaler.fit_transform(feature)
```

```python
# Apply Min-Max Scaling
minmax_scaler = MinMaxScaler()
df_minmax = df.copy()
df_minmax["LotArea"] = minmax_scaler.fit_transform(feature)
```

```python
# Apply Robust Scaling
robust_scaler = RobustScaler()
df_robust = df.copy()
df_robust["LotArea"] = robust_scaler.fit_transform(feature)
```

```python
# Compare the results
print("Original Data:")
print(df.head())

print("\nStandard Scaled:")
print(df_standard.head())

print("\nMin-Max Scaled:")
print(df_minmax.head())

print("\nRobust Scaled:")
print(df_robust.head())
```

## **Encoding**

```python
# Upload dataset file:
from google.colab import files

uploaded = files.upload()
```

```python
import pandas as pd
# Load the dataset
df = pd.read_excel("encoding_data.xlsx")
```

```python
# Encode the categorical columns
df_encoded = pd.get_dummies(
    df,
    columns=["item_category", "Gender"],
    dtype=int
)

# Encode the binary target column
df_encoded["Purchased"] = df_encoded["Purchased"].map({
    "No": 0,
    "Yes": 1
})
```

```python
# Display the encoded data
print(df_encoded)
```
